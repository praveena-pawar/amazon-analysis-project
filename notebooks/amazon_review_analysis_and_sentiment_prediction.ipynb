```python
df = df[['name', 'brand', 'categories', 'reviews.rating',
         'reviews.text', 'reviews.title', 'reviews.numHelpful']]
df
```

```python
df = df[df['reviews.rating'].notna()]
df['reviews.numHelpful'] = df['reviews.numHelpful'].fillna(0)
```

```python
df['reviews.title'] = df['reviews.title'].fillna("")
```

```python
# Analyze distribution of product ratings to understand overall sentiment in the dataset
import matplotlib.pyplot as plt

df['reviews.rating'].value_counts().sort_index().plot(kind='bar')
plt.title("Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Count")
plt.show()
```

```python
# Identify top 10 brands with highest number of reviews to understand brand dominance

df['brand'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Brands by Number of Reviews")
plt.xlabel("Brand")
plt.ylabel("Number of Reviews")
plt.show()
```

```python
# Analyze distribution of review lengths to understand how detailed user reviews are

df['review_length'] = df['reviews.text'].apply(len)

df['review_length'].hist()
plt.title("Review Length Distribution")
plt.xlabel("Review Length (Number of Characters)")
plt.ylabel("Number of Reviews")
plt.show()
```

```python
# Analyze relationship between product ratings and average review length to understand user behavior patterns

df.groupby('reviews.rating')['review_length'].mean().plot(kind='bar')
plt.title("Average Review Length by Rating")
plt.xlabel("Rating")
plt.ylabel("Average Review Length")
plt.show()
```

```python
# Create binary label: 1 = good (rating >= 4), 0 = bad

df['label'] = df['reviews.rating'].apply(lambda x: 1 if x >= 4 else 0)
```

```python
# Combine title and text for better understanding

df['combined_text'] = df['reviews.title'] + " " + df['reviews.text']
```

```python
# Take only main category

df['categories'] = df['categories'].apply(lambda x: x.split(',')[0])
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(max_features=1000, stop_words='english')

X_text = tfidf.fit_transform(df['combined_text'])
```

```python
from scipy.sparse import hstack

X_num = df[['review_length', 'reviews.numHelpful']]

X = hstack([X_text, X_num])
y = df['label']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier()
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, classification_report

y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(max_iter=1000, class_weight='balanced')
lr.fit(X_train, y_train)
```
